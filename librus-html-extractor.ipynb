```python
#Import of Helena's grades

import pandas as pd

file_path = 'G:\Mój dysk\librus-extractor\input-data\Helena.html'

try:
    # Reads the HTML file and searches for a table containing the matching string. 
    # (Using lower case 'bieżące' as Librus often capitalizes only the first word)
    tables = pd.read_html(file_path, match='Oceny bieżące')
    
    if tables:
        # pandas returns a list of matching tables, we select the first one
        df_oceny = tables[0]
        
        print("Table successfully found:")
        print(df_oceny.head())
        
        # Optional: Save the scraped table to a CSV file
        df_oceny.to_csv('G:\Mój dysk\librus-extractor\output-data\Helena.csv', index=False, encoding='utf-8-sig')
    else:
        print("Table was found but returned empty.")

except ValueError:
    print("Could not find a table matching 'Oceny bieżące'.")
```
